## Introducción

En este documento se presenta el análisis y modelado cinemático de un robot Scara, con enfoque en la descripción matemática de su movimiento en una dimensión 2D. El objetivo principal es establecer y explicar las relaciones entre el espacio de las articulaciones del robot (postura) y el espacio de la tarea (pose del efector), mediante la deducción de los modelos directos e inversos.


A lo largo del documento se abordan los siguientes tópicos:

-  **Cinemática de la Postura**: Se define el vector de pose $^0 \xi_p$ para describir la posición y orientación del efector, y el vector de postura $^0 \xi_p (q)$ para la disposición de los eslabones. Se utiliza el método de matrices de transformación homogénea para obtener el modelo directo, aplicando las restricciones de movimiento específicas del robot.
-  **Cinemática de Velocidades y Aceleraciones**: Se deriva el modelo de velocidades a través de la matriz Jacobiana $J(q)$, la cual permite proyectar las velocidades tangenciales y angulares de cada eslabón respecto al sistema inercial. Asimismo, se extiende el análisis a las aceleraciones mediante la segunda derivada del vector de pose.
-  **Cinemática Inversa**: Se propone un método algebraico para resolver el problema inverso de la postura, permitiendo determinar la configuración necesaria de los eslabones para alcanzar una pose deseada. Este proceso incluye la simplificación del sistema mediante la omisión estratégica del efecto del tercer eslabón para reducir el número de incógnitas.
## Modelo cinemático directo del robot Scara
### Modelo cinemático directo de la postura




**Figura 1.** Diagrama cinemático para el análisis de la posición del robot Scara.


Para obtener el modelo cinemático de la postura del robot Scara, se parte de la siguiente ecuación:


$$\begin{array}{l} \newline ^0 \xi_p =^0 \xi_p (q)\newline \newline \textrm{Donde:}\newline ^0 \xi_p ~~\textrm{es}\;\textrm{el}\;\textrm{vector}\;\textrm{que}\;\textrm{describe}\;\textrm{la}\;\textrm{pose}\;\textrm{que}\;\textrm{tiene}\;\textrm{el}\;\textrm{efector}\;\textrm{del}\;\textrm{robot;}\;\textrm{y}\newline ^0 \xi_p (q)~~\textrm{es}\;\textrm{el}\;\textrm{vector}\;\textrm{que}\;\textrm{describe}\;\textrm{la}\;\textrm{postura}\;\textrm{que}\;\textrm{debe}\;\textrm{adoptar}\;\textrm{el}\;\textrm{robot}\;\textrm{para}\;\textrm{alcanzar}\;\textrm{la}\;\textrm{pose}\;\textrm{del}\;\textrm{efector.}\newline  \end{array}$$

Además, el vector de pose se define como:


$$\begin{array}{l} \newline ^0 \xi_p =\binom{^0 P_p }{^0 \theta_p }\newline \newline \textrm{Donde:}\newline ^0 P_p ~~\textrm{es}\;\textrm{el}\;\textrm{vector}\;\textrm{de}\;\textrm{posición}\;\textrm{del}\;\textrm{efector;}\;\textrm{y}\newline ^0 \theta_p ~~\textrm{es}\;\textrm{el}\;\textrm{vector}\;\textrm{de}\;\textrm{orientación}\;\textrm{del}\;\textrm{efector.}\newline  \end{array}$$

De la misma manera, la definición de los vectores que conforman al vector de pose es la siguiente:

 $ $ \begin{array}{l} \newline ^0 P_p =\left(\begin{array}{c} ^0 x_p \newline ^0 y_p \newline ^0 z_p  \end{array}\right)~~\textrm{donde}\;\textrm{"x",}\;\textrm{"y"}\;\textrm{y}\;\textrm{"z"}\;\textrm{son}\;\textrm{las}\;\textrm{coordenadas}\;\textrm{del}\;\textrm{vector}\;\textrm{de}\;\textrm{posición}\;\textrm{del}\;\textrm{efector}\\
\\
\textrm{y}\\
\\
^0 \theta_p =\left(\begin{array}{c} ^0 \gamma_p \newline ^0 \beta_p \newline ^0 \alpha_p  \end{array}\right)~~\textrm{donde}\;\textrm{"}\gamma \textrm{",}\;\textrm{"}\beta \textrm{"}\;\textrm{y}\;\textrm{"}\alpha \textrm{"}\;\textrm{son}\;\textrm{las}\;\textrm{rotaciones}\;\textrm{de}\;\textrm{los}\;\textrm{ejes}\;\textrm{del}\;\textrm{sistema}\;\textrm{referencial}\;\textrm{relativo}\;\textrm{del}\;\textrm{efector.}\\
\end{array} $ $

De esta manera, el vector de pose se puede escribir como:

 $ $ \begin{array}{l} \newline ^0 \xi_p =\left(\begin{array}{c} ^0 x_p \newline ^0 y_p \newline ^0 z_p \newline ^0 \gamma_p \newline ^0 \beta_p \newline ^0 \alpha_p  \end{array}\right)\\
\end{array} $ $

Sin embargo, los eslabones tienen restringido el movimiento en "z", " $\gamma$ " y " $\beta$ ", véase la figura 1. Debido a esto, se puede hacer la siguiente simplificación:

 $ $ \begin{array}{l} \newline ^0 \xi_p =\left(\begin{array}{c} ^0 x_p \newline ^0 y_p \newline 0\newline 0\newline 0\newline ^0 \alpha_p  \end{array}\right)\to ^0 \xi_p =\left(\begin{array}{c} ^0 x_p \newline ^0 y_p \newline ^0 \theta_p  \end{array}\right)\\
\\
\end{array} $ $




**Figura 2.** Representación del vector de pose sobre el diagrama cinemático del robot Scara.


Por otro lado, el vector $^0 \xi_p (q)$ describe la orientación que debe adoptar cada uno de los eslabones que conforman al robot para alcanzar la pose del efector. La función "q" representa la orientación de cada eslabón según su pose relativa a la articulación, o grado de libertad, que es capaz de provocar cambios en su descripción espacial \[2\]:

 $ $ \begin{array}{l} \newline q=\left(\begin{array}{c} q_1 \newline \cdot \newline \cdot \newline \cdot \newline q_n  \end{array}\right)~~\textrm{siendo}\,q_1 \,\textrm{hasta}\,q_n \,\textrm{la}\;\textrm{orientación}\;\textrm{realtiva}\;\textrm{al}\;\textrm{sistema}\;\textrm{referencial}\;\textrm{de}\;\textrm{la}\;\textrm{primera}\;\textrm{articulación,}\;\textrm{o}\;\textrm{grado}\;\textrm{de}\;\textrm{libertad,}\;\textrm{de}\;\textrm{cada}\;\textrm{uno}\;\textrm{de}\;\textrm{los}\;\textrm{eslabones}\;\textrm{que}\;\textrm{conforman}\;\textrm{al}\;\textrm{robot.}\\
\end{array} $ $

De esta manera, el vector de postura $^0 \xi_p (q)$ puede entenderse como la dispoción de los eslabones del robot para alcanzar la pose del efector.





**Figura 3.** Representación de los vectores de la función "q" que describen la postura del robot.


Para describir la variación de cada sistema referencial relativo a las articulaciones, o grados de libertad, respecto con el sistema inercial se utilizan las matrices de transformación homogénea:

 $ $ \begin{array}{l} \newline ^i T_j =\left(\begin{array}{cc} ^i R_j (\theta ) & ^i P_j \newline 0^T  & 1 \end{array}\right)\\
\\
\textrm{Donde:}\\
^i T_j ~~\textrm{es}\;\textrm{la}\;\textrm{matriz}\;\textrm{de}\;\textrm{transformación}\;\textrm{homogénea;}\\
^i R_j (\theta )~~\textrm{es}\;\textrm{la}\;\textrm{matriz}\;\textrm{de}\;\textrm{rotación}\;\textrm{que}\;\textrm{describe}\;\textrm{la}\;\textrm{orientación}\;\textrm{del}\;\textrm{sistema}\;\textrm{"j"}\;\textrm{respecto}\;\textrm{al}\;\textrm{sistema}\;\textrm{"i";}\\
^i P_j ~~\textrm{es}\;\textrm{el}\;\textrm{vector}\;\textrm{de}\;\textrm{posición}\;\textrm{de}\;\textrm{lo}\;\textrm{que}\;\textrm{se}\;\textrm{quiere}\;\textrm{describir}\;\textrm{en}\;\textrm{el}\;\textrm{sistema}\;\textrm{"j"}\;\textrm{que}\;\textrm{pertenece}\;\textrm{al}\;\textrm{sistema}\;\textrm{"i";}\\
0^T ~~\textrm{es}\;\textrm{el}\;\textrm{vector}\;\textrm{de}\;\textrm{perspectiva;}\;\textrm{y}\\
1~~\textrm{es}\;\textrm{la}\;\textrm{escala.}\\
\\
\end{array} $ $
#### Cálculo de la matriz de transformación homogénea generalizada

```````matlab
%Symbolic funtion definition
syms Tij(x_i_j,y_i_j,z_i_j,gi_j,bi_j,ai_j)

%Homogeneous transform
Tij(x_i_j,y_i_j,z_i_j,gi_j,bi_j,ai_j) = [cos(ai_j)*cos(bi_j) cos(ai_j)*sin(bi_j)*sin(gi_j)-sin(ai_j)*cos(gi_j) sin(ai_j)*sin(gi_j)+cos(ai_j)*sin(bi_j)*cos(gi_j) x_i_j; sin(ai_j)*cos(bi_j) cos(ai_j)*cos(gi_j)+sin(ai_j)*sin(bi_j)*sin(gi_j) sin(ai_j)*sin(bi_j)*cos(gi_j)-cos(ai_j)*sin(gi_j) y_i_j; -sin(bi_j) cos(bi_j)*sin(gi_j) cos(bi_j)*cos(gi_j) z_i_j; 0 0 0 1]
```````


Sin embargo, aplicando las restricciones de movimiento observables en la figura 1, que corresponden a nuestro caso de estudio; la matriz resulta como se muestra a continuación:

 $ $ \begin{array}{l} \newline ^i T_j =\left(\begin{array}{cccc} \cos ^i \theta_j  & -\sin ^i \theta_j  & 0 & ^i x_j \newline \sin ^i \theta_j  & \cos ^i \theta_j  & 0 & ^i y_j \newline 0 & 0 & 1 & 0\newline 0 & 0 & 0 & 0 \end{array}\right)\\
\end{array} $ $

Al observar el diagrama de la figura 3, se puede establecer que se necesita una matriz de transformación homogénea que describa el sistema 1 en el sistema inercial, el sistema 2 en el sistema 1, y el sistema 3 en el sistema 2 para poder describir todos con respecto al inercial:

 $ $ \begin{array}{l} \newline ^0 T_p =^0 T_1 ^1 T_2 ^2 T_3 \to ^0 T_p =\left(\begin{array}{cc} ^0 R_1 (\theta ) & ^0 P_1 \newline 0^T  & 1 \end{array}\right)\left(\begin{array}{cc} ^1 R_2 (\theta ) & ^1 P_2 \newline 0^T  & 1 \end{array}\right)\left(\begin{array}{cc} ^2 R_3 (\theta ) & ^2 P_3 \newline 0^T  & 1 \end{array}\right)\to ^0 T_p =\left(\begin{array}{cc} ^0 R_p (\theta ) & ^0 P_p \newline 0^T  & 1 \end{array}\right)\\
\end{array} $ $

Una vez calculada la matriz de transformación homogénea que describa la posición de los 3 eslabones respecto al sistema inercial, la información del vector de posición $^0 P_p$ se utiliza para definir el vector de postura del robot:

 $ $ \begin{array}{l} \newline ^0 \xi_p (q)=\left(\begin{array}{c} ^0 x_p \newline ^0 y_p \newline ^0 \theta_1 +^1 \theta_2 +^2 \theta_3  \end{array}\right)\\
\end{array} $ $
#### Cálculo del modelo

```````matlab
% Definicion los parámetros
syms x_O_1 y_O_1  x_1_2 x_2_3 x_3_P
% Definicion de las variables
syms theta_O_1 theta_1_2 theta_2_3

TO_1 = Tij(x_O_1,y_O_1,0,0,0,theta_O_1)
```````

```````matlab
T_1_2 = Tij(x_1_2,0,0,0,0,theta_1_2)
```````

```````matlab
T_2_3 = Tij(x_2_3,0,0,0,0,theta_2_3)
```````

```````matlab
T_3_P = Tij(x_3_P,0,0,0,0,0)
```````

```````matlab
T_O_P = simplify(TO_1*T_1_2*T_2_3*T_3_P)
```````

```````matlab

syms x_O_P y_O_P theta_O_P
xi_O_P = [x_O_P; y_O_P; theta_O_P]
```````

```````matlab
xi_O_P_q = [T_O_P(1,4);T_O_P(2,4); theta_O_1+theta_1_2+theta_2_3]
```````

### Modelos cinemático directo de las velocidades




**Figura 4.** Diagrama de las trayectorias y Jacobianos de velocidades de los eslabones del robot Scara.


Para obtener el modelo de las velocidades simplemente se deriva la ecuación del modelo cinemático de la postura:


$$\begin{array}{l} \newline ^0 {\dot{\xi} }_p =J(q)\dot{q} \newline \newline \textrm{Donde:}\newline ^0 {\dot{\xi} }_p ~~\textrm{es}\;\textrm{el}\;\textrm{vector}\;\textrm{de}\;\textrm{la}\;\textrm{velocidad}\;\textrm{final}\;\textrm{del}\;\textrm{efector}\;\textrm{respecto}\;\textrm{al}\;\textrm{sistema}\;\textrm{inercial;}\newline J(q)~~\textrm{es}\;\textrm{la}\;\textrm{matriz}\;\textrm{de}\;\textrm{derivadas}\;\textrm{parciales}\;\textrm{de}\;\textrm{la}\;\textrm{pose}\;\textrm{de}\;\textrm{cada}\;\textrm{eslabón}\;\textrm{del}\;\textrm{robot;}\;\textrm{y}\newline \dot{q} ~~\textrm{es}\;\textrm{la}\;\textrm{derivada}\;\textrm{de}\;\textrm{la}\;\textrm{función}\;\textrm{"q".}\newline  \end{array}$$

La derivada del vector de pose $^0 \xi_p$ equivale a las proyecciones sobre los ejes "x" y "y" de la velocidad tangencial y a la velocidad angular sobre el eje "z" (Yaw) del efector:

 $ $ \begin{array}{l} \newline ^0 {\dot{\xi} }_p =\frac{d}{dt}^0 \xi_p =\left(\begin{array}{c} ^0 {\dot{x} }_p \newline ^0 {\dot{y} }_p \newline ^0 {\dot{\theta} }_p  \end{array}\right)\\
\end{array} $ $

La matriz de derivadas parciales y la función $\dot{q}$ se obtienen a partir de la derivada del vector de postura $^0 {\dot{\xi} }_p$:

 $ $ \begin{array}{l} \newline ^0 {\dot{\xi} }_p (q)=\frac{\partial }{\partial \;^0 \theta_1 }^0 \xi_p ^0 {\dot{\theta} }_1 +\frac{\partial }{\partial \;^1 \theta_2 }^0 \xi_p ^1 {\dot{\theta} }_2 +\frac{\partial }{\partial \;^2 \theta_3 }^0 \xi_p ^2 {\dot{\theta} }_3 \newline \newline \textrm{Donde:}\newline \frac{\partial }{\partial \;^0 \theta_1 }^0 \xi_p +\frac{\partial }{\partial \;^1 \theta_2 }^0 \xi_p +\frac{\partial }{\partial \;^2 \theta_3 }^0 \xi_p =J(q);~~\textrm{y}\newline \left(\begin{array}{c} ^0 {\dot{\theta} }_1 \newline ^1 {\dot{\theta} }_2 \newline ^2 {\dot{\theta} }_3  \end{array}\right)=\dot{q} \\
\end{array} $ $

Además, la matriz de derivadas parciales $J(q)$, o jacobiano, equviale a las proyecciones de la velocidad tangencial y angular en cada uno de los eslabones, es decir; del vector de postura $^0 \xi_p (q)$:

 $ $ \begin{array}{l} \newline J(q)=\left(\begin{array}{ccc} Jv_1  & Jv_2  & Jv_3 \newline J\omega_1  & J\omega_2  & J\omega_3  \end{array}\right)\\
\end{array} $ $

Donde los vectores $Jv_1 ,\;Jv_2 \;\textrm{y}\;Jv_3$ son la suma de las proyecciones de las velocidades tangenciales realtivas al sistema referencial de la primera articulaicón, o grado de libertad, de cada eslabón del robot a partir del eslabon que el subíndice indique:


$$\begin{array}{l} \newline Jv_1 =\binom{^1 {\dot{x} }_2 +^2 {\dot{x} }_3 +^3 {\dot{x} }_p }{^1 {\dot{y} }_2 +^2 {\dot{y} }_3 +^3 {\dot{y} }_p };~~Jv_2 =\binom{^2 {\dot{x} }_3 +^3 {\dot{x} }_p }{^2 {\dot{y} }_3 +^3 {\dot{y} }_p }~~\textrm{y}~~Jv_3 =\binom{^3 {\dot{x} }_p }{^3 {\dot{y} }_p }\newline  \end{array}$$

Asimismo, los vectores $J\omega_1 ,\;J\omega_2 \;\textrm{y}\;J\omega_3$ son la suma de las velocidades angulares siguendo la misma lógica que las velocidades tangenciales:

 $ $ \begin{array}{l} \newline J\omega_1 =\left(\begin{array}{c} ^0 {\dot{\theta} }_1 +^1 {\dot{\theta} }_2 +^2 {\dot{\theta} }_3  \end{array}\right);~~J\omega_2 =\left(\begin{array}{c} ^1 {\dot{\theta} }_2 +^2 {\dot{\theta} }_3  \end{array}\right)~~\textrm{y}~~J\omega_3 =\left(\begin{array}{c} ^2 {\dot{\theta} }_3  \end{array}\right)\\
\end{array} $ $
#### Cálculo del modelo

```````matlab
q_dot = [theta_dot_O_1; theta_dot_1_2; theta_dot_2_3]
```````

```````matlab
J_theta = simplify(jacobian(xi_O_P_q,[theta_O_1,theta_1_2,theta_2_3]))
```````

### Modelo cinemático directo de las aceleraciones

Para obtener las aceleraciones, se deriva nuevamente:


$$\begin{array}{l} \newline ^0 {\ddot{\xi} }_p =J(q)\ddot{q} +\dot{J} (q)\dot{q} \newline  \end{array}$$

La segunda derivada del vector de pose equivale a las proyecciones de la aceleracion tangencial del efector:

 $ $ \begin{array}{l} \newline ^0 {\ddot{\xi} }_p =\left(\begin{array}{c} ^0 {\ddot{x} }_p \newline ^0 {\ddot{y} }_p \newline ^0 {\ddot{\theta} }_p  \end{array}\right)\\
\end{array} $ $

A su vez, la segunda derivada del vector de postura resulta en la derivada de la multiplicación del jacobiano en función de "q" y la propia función "q", es decir, la derivada de un producto de funciones:

 $ $ \begin{array}{l} \newline ^0 {\ddot{\xi} }_p (q)=\frac{d}{dt}J(q)\dot{q} \to ^0 {\ddot{\xi} }_p (q)=J(q)\frac{d}{dt}\dot{q} +\frac{d}{dt}J(q)\;\dot{q} \newline \newline \textrm{Donde:}\newline J(q)\frac{d}{dt}\dot{q} =J(q)\ddot{q} \,;~~\textrm{y}\newline \frac{d}{dt}J(q)\;\dot{q} =\left(\begin{array}{c} \frac{\partial^2 }{\partial \;^0 \theta_1^2 }^0 \xi_p +\frac{\partial^2 }{\partial \;^1 \theta_2^2 }^0 \xi_p +\frac{\partial^2 }{\partial \;^2 \theta_3^2 }^0 \xi_p  \end{array}\right)\dot{q} \\
\end{array} $ $
#### Cálculo del modelo

```````matlab

syms theta_ddot_O_1 theta_ddot_1_2 theta_ddot_2_3
q_ddot = [theta_ddot_O_1; theta_ddot_1_2; theta_ddot_2_3]
```````

```````matlab
J_dot_theta = simplify(diff(J_theta,theta_O_1)*theta_dot_O_1 + diff(J_theta,theta_1_2)*theta_dot_1_2 + diff(J_theta,theta_2_3)*theta_dot_2_3)
```````

```````matlab
xi_ddot_O_P = simplify(J_theta*q_ddot + J_dot_theta*q_dot)
```````

```````matlab
simplify(J_theta*q_ddot )
```````

```````matlab
simplify(J_dot_theta*q_dot)
```````

 $ $ \begin{array}{l} \newline \left(\begin{array}{c} ^0 {\ddot{x} }_p \newline ^0 {\ddot{y} }_p \newline ^0 {\ddot{\theta} }_p  \end{array}\right)\\
